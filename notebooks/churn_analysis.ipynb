```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

df["TotalCharges"].isnull().sum()
```

```python
categorical_columns = df.select_dtypes(include="str").columns

for col in categorical_columns:
    print("\n" + "="*40)
    print(col)
    print("="*40)
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    spaces = df[col].astype(str).str.strip().ne(df[col].astype(str)).sum()
    print(col, ":", spaces)
```

```python
df["SeniorCitizen_Status"] = np.where(
    df["SeniorCitizen"] == 1,
    "Senior Citizen",
    "Non-Senior Citizen"
)
```

```python
for col in categorical_columns:
    if col != "customerID":
        print(f"\n{col}:")
        print(sorted(df[col].unique()))
```

```python
df[["SeniorCitizen", "SeniorCitizen_Status"]].drop_duplicates()
```

```python
df["Churn_Flag"] = np.where(
    df["Churn"] == "Yes",
    1,
    0
)
```

```python
df["Churn_Flag"].value_counts()
```

```python
df["Customer_Status"] = np.where(
    df["Churn"] == "Yes",
    "Churned",
    "Retained"
)
```

```python
df["Customer_Status"].value_counts()
```

```python
df["Tenure_Group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 36, 48, 60, 72],
    labels=[
        "0-12 Months",
        "13-24 Months",
        "25-36 Months",
        "37-48 Months",
        "49-60 Months",
        "61-72 Months"
    ]
)
```

```python
df["Tenure_Group"].value_counts(sort=False)
```

```python
df["Revenue_Lost_Monthly"] = np.where(
    df["Churn"] == "Yes",
    df["MonthlyCharges"],
    0
)
```

```python
df["Annual_Revenue_Lost"] = (
    df["Revenue_Lost_Monthly"] * 12
)
```

```python
print("Total customers:", len(df))

print("\nChurn flag:")
print(df["Churn_Flag"].value_counts())

print("\nCustomer status:")
print(df["Customer_Status"].value_counts())

print("\nTenure groups:")
print(df["Tenure_Group"].value_counts(sort=False))

print("\nMonthly revenue associated with churn:")
print(df["Revenue_Lost_Monthly"].sum())

print("\nAnnualized equivalent:")
print(df["Annual_Revenue_Lost"].sum())

print("\nMissing values in new features:")
print(
    df[
        [
            "Churn_Flag",
            "Customer_Status",
            "Tenure_Group",
            "Revenue_Lost_Monthly",
            "Annual_Revenue_Lost"
        ]
    ].isnull().sum()
)
```

```python
churn_rate = df["Churn_Flag"].mean() * 100

print(f"Overall churn rate: {churn_rate:.2f}%")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

churn_counts = df["Churn"].value_counts()

print(churn_counts)
print("\nChurn percentage:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))
```

```python
sns.countplot(data=df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
contract_analysis = (
    df.groupby("Contract", observed=True)
      .agg(
          Total_Customers=("customerID", "count"),
          Churned_Customers=("Churn_Flag", "sum"),
          Churn_Rate=("Churn_Flag", "mean")
      )
      .reset_index()
)

contract_analysis["Churn_Rate"] *= 100

contract_analysis = contract_analysis.sort_values(
    "Churn_Rate", ascending=False
)

print(contract_analysis.round(2))
```

```python
sns.barplot(
    data=contract_analysis,
    x="Contract",
    y="Churn_Rate"
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
tenure_analysis = (
    df.groupby("Tenure_Group", observed=False)
      .agg(
          Total_Customers=("customerID", "count"),
          Churned_Customers=("Churn_Flag", "sum"),
          Churn_Rate=("Churn_Flag", "mean")
      )
      .reset_index()
)

tenure_analysis["Churn_Rate"] *= 100

print(tenure_analysis.round(2))
```

```python
sns.lineplot(
    data=tenure_analysis,
    x="Tenure_Group",
    y="Churn_Rate",
    marker="o"
)

plt.title("Churn Rate Across Customer Tenure")
plt.xlabel("Tenure Group")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
def analyze_churn_by(column):
    result = (
        df.groupby(column, observed=True)
          .agg(
              Total_Customers=("customerID", "count"),
              Churned_Customers=("Churn_Flag", "sum"),
              Churn_Rate=("Churn_Flag", "mean")
          )
          .reset_index()
    )

    result["Churn_Rate"] *= 100

    return result.sort_values(
        "Churn_Rate", ascending=False
    )
```

```python
internet_analysis = analyze_churn_by("InternetService")
print(internet_analysis.round(2))
```

```python
payment_analysis = analyze_churn_by("PaymentMethod")
print(payment_analysis.round(2))
```

```python
sns.barplot(
    data=internet_analysis,
    x="InternetService",
    y="Churn_Rate"
)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=payment_analysis,
    x="PaymentMethod",
    y="Churn_Rate"
)

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
total_monthly_charges = df["MonthlyCharges"].sum()

monthly_revenue_lost = df["Revenue_Lost_Monthly"].sum()

revenue_share = (
    monthly_revenue_lost / total_monthly_charges
) * 100

print(f"Total monthly charges: ${total_monthly_charges:,.2f}")
print(f"Charges associated with churned customers: ${monthly_revenue_lost:,.2f}")
print(f"Share of total monthly charges: {revenue_share:.2f}%")
```

```python
revenue_comparison = pd.DataFrame({
    "Customer Status": ["Retained", "Churned"],
    "Monthly Charges": [
        df.loc[df["Churn"] == "No", "MonthlyCharges"].sum(),
        df.loc[df["Churn"] == "Yes", "MonthlyCharges"].sum()
    ]
})

sns.barplot(
    data=revenue_comparison,
    x="Customer Status",
    y="Monthly Charges"
)

plt.title("Monthly Charges by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Monthly Charges ($)")
plt.show()
```
